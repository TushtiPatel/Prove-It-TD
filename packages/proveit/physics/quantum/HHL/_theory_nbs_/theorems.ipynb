For Adjoint

```python
adjoint_of_ket = Forall(x,
                        Equals(Adj(Ket(x)), Bra(x)))
```

```python
adj_scalar = Forall(x,
                        Equals(Adj(x), Conjugate(x)), domain=Complex)
```

```python
adj_of_adj = Forall(x,
                        Equals(Adj(Adj(x)), x))
```

```python
adjoint_of_linear_combination = Forall(Hspace,
    Forall((x, y),
        Forall((a, b),
            Equals(Adj(VecAdd(ScalarMult(a, Ket(x)), ScalarMult(b, Ket(y)))),
                    VecAdd(ScalarMult(Conjugate(a), Bra(x)),
                    ScalarMult(Conjugate(b),Bra(y) ))), domain=Complex
                ), domain=Hspace
            ),
        )
```

```python
adjoint_of_linear_combination_n_terms = Forall(n, 
                                               Forall(Hspace,
                                                        Forall((a,x),
                                                             Implies(Forall(i, And( InSet(_a_i,Complex), InSet(Ket(_x_i),Hspace) ),domain = Interval(zero,n) ),
                                                             Equals(Adj(VecSum(i,ScalarMult(_a_i,Ket(_x_i)), domain=Interval(zero,n))),
                                                                  VecSum(i,ScalarMult(Conjugate(_a_i),Bra(_x_i)), domain=Interval(zero,n)) ) ),
                                                        ),
                                                 condition = IsHilbertSpace(Hspace)),
                                               domain = Natural)



# forall_{n in N} forall_{a, x} Implies((forall_{i in 0 .. n} a_i in C) and (forall_{i in 0 .. n} x_i in H) 
 
```

```python
adjoint_vecsum = Forall((a,b),
                        Forall(x,
                               Equals(Adj(VecSum(i,Function(x,i), domain = Interval(a,b))),VecSum(i,Adj(Function(x,i)),  domain = Interval(a,b)))
                               ),
                        domain=Natural)
```

```python
adj_qmult_unary = Forall(
    A,
    Equals(
        Adj(Qmult(A)),
        Qmult(Adj(A))
    ),
    conditions=[IsProperQmult(A)]
)
```

```python
adj_qmult_binary = Forall((A,B), Equals(Adj(Qmult(A,B)), Qmult(Adj(B),Adj(A))), condition=IsProperQmult(Qmult(A, B)) )

```

```python
adj_linmaps = Forall(
    (V, W, A),
    Implies(
        InSet(A, LinMaps(V, W)),
        InSet(Adj(A), LinMaps(W, V))
    )
)
```

```python
adj_linmaps_converse = Forall(
    (V, W, A),
    Implies(
        InSet(Adj(A), LinMaps(V, W)),
        InSet(A, LinMaps(W, V))
    )
)
```
