```python
import proveit
# Prepare this notebook for defining the common expressions of a theory:
%common_expressions_notebook # Keep this at the top following 'import proveit'.

from proveit import (Literal, Variable, ExprRange, ExprArray, VertExprArray,
                     ConditionalSet, Conditional, ExprTuple, Function)
from proveit.physics.quantum.QPE.phase_est_ops import ModAdd, SubIndexed

from proveit.physics.quantum import Ket

from proveit.linear_algebra import VecSum, ScalarMult


from proveit.numbers import zero, one, two, three, Exp, subtract, Interval, frac, Complex, Add

from proveit import j, k, x, i, a

from proveit.logic import CartExp
```

```python
_u = Literal('u', latex_format=r'u')
# basis state of the input vector
```

```python
_u_j_ket = Ket(SubIndexed(_u, j))
```

```python
_beta = Literal('beta', latex_format=r'\beta')
# coefficient of the basis state of the input vector
```

```python
_beta_j = SubIndexed(_beta,j)
```

```python
_inp_state = VecSum(j, ScalarMult(_beta_j,_u_j_ket), domain = Interval(zero, subtract(Exp(two,_nb),one) ) )
```

```python
_alpha = Literal('alpha', latex_format=r'\alpha')
# coefficient for clock register qubits
```

```python
_alpha_kj = Function(_alpha,ExprTuple(k,j))
```

```python
_clock_state = VecSum(k, ScalarMult(_alpha_kj,Ket(k)), domain = Interval(zero, subtract(Exp(two,_nt),one) ) )
```
